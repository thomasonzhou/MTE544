```python
def plot_laserscan(data, step_rad=0.01749303564429283, offset=math.pi / 2.0, title="Plot"):
    coords = []
    for i in range(360):
        true_angle = (offset + i * step_rad) % (math.pi * 2.0)
        distance = float(data[i])
        if distance != float("inf"):
            theta = true_angle
            x = distance * math.cos(theta)
            y = distance * math.sin(theta)
            coords.append((x, y))

    # print(coords)

    plt.scatter([coord[0] for coord in coords], [coord[1] for coord in coords])
    plt.title(title)
    plt.xlabel("X (meters)")
    plt.ylabel("Y (meters)")
```
